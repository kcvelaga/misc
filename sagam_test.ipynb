```python
import requests
import json
```

```python
def process_text(text, service):
    
    api_endpoint = f"https://sangam.learnpunjabi.org/SindhiTransliteration.asmx/{service}"
    
    text_ip = {
        "input": text # good morning
    }

    headers = {
        "User-Agent": "KCVelaga-test (@kcvelaga)",
        "Accept": "application/json, text/javascript, */*; q=0.01",
        "Accept-Language": "en-US,en;q=0.5",
        "Content-Type": "application/json; charset=utf-8",
        "X-Requested-With": "XMLHttpRequest",
        "Sec-Fetch-Dest": "empty",
        "Sec-Fetch-Mode": "cors",
        "Sec-Fetch-Site": "same-origin",
        "Sec-GPC": "1",
        "Priority": "u=0"
    }

    response = requests.post(api_endpoint, data=json.dumps(text_ip), headers=headers)

    return(response.text)
```

```python
json.loads(process_text("ہیلو، صبح بخیر", "Urdu2Hindi"))['d']
```

```python
json.loads(process_text("ਹੈਲੋ ਸ਼ੁਭ ਸਵੇਰ", "Gurmukhi2Shahmukhi"))['d']
```
